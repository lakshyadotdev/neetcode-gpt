##### Gradient Descent
weights ki most apporpriate values dhundhna, by iterating and updating learning rate but yaha constant h
f(x) = x<sup>2</sup>

```python
def minimize(learning_rate:float,iterations:int,init:float):
    x = init
    for _ in range(iterations):
        slope = 2*x
        x -= learning_rate*slope
    return round(x,5)
minimize(0.05, 100, 5)
```

#### Activation Functions 
isse pta chalta h ki agle functions me kitne confidence ke sath is neuron ko bhejna h, aise to kitne baar bhi krlo wohi compute hoga, to inputs*weights me bias ko add krta h

```python
def sigmoid(z: NDArray[np.float64]) -> NDArray[np.float64]:
        #converts it into 0 -> 1
        return 1 / (1 + np.exp(-z))

def relu(z: NDArray[np.float64]) -> NDArray[np.float64]:
        # positive deta h
    return np.maximum(0,z)
sigmoid(np.array([1,2,32]))

```

#### Softmax (activation ka implementaion)

```python
def softmax(z: NDArray[np.float64]) -> NDArray[np.float64]:
        return np.round(np.exp(z- np.max(z)) / np.sum(np.exp(z- np.max(z))),4)
softmax(np.array([1,27,32]))
```
